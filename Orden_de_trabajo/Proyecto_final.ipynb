```python
class Producto:
    def __init__(self, nombre, precio, cantidad):
        self.nombre = nombre
        self.precio = precio
        self.cantidad = cantidad

    def disponible(self):
        return self.cantidad > 0

    def reducir_cantidad(self):
        self.cantidad -= 1
```

```python
class Compra:
    def __init__(self, producto, dinero_ingresado):
        self.producto = producto
        self.dinero_ingresado = dinero_ingresado
        self.cambio = dinero_ingresado - producto.precio

    def mostrar_resumen(self):
        print(f"Compraste: {self.producto.nombre}")
        print(f"Precio: ${self.producto.precio}")
        print(f"Pagaste: ${self.dinero_ingresado}")
        print(f"Tu cambio es: ${self.cambio}")
```

```python
class MaquinaExpendedora:
    def __init__(self):
        self.productos = []

    def agregar_producto(self, producto):
        self.productos.append(producto)

    def mostrar_productos(self):
        print("\n--- Productos disponibles ---")
        for i, producto in enumerate(self.productos):
            if producto.disponible():
                print(f"{i+1}. {producto.nombre} - ${producto.precio}")
            else:
                print(f"{i+1}. {producto.nombre} - AGOTADO")

    def comprar(self, numero_producto, dinero_ingresado):
        producto = self.productos[numero_producto - 1]

        if not producto.disponible():
            print("Lo siento, ese producto está agotado.")
            return

        if dinero_ingresado < producto.precio:
            print(f"Dinero insuficiente. El producto cuesta ${producto.precio}")
            return


        compra = Compra(producto, dinero_ingresado)
        producto.reducir_cantidad()
        compra.mostrar_resumen()
```

```python
maquina = MaquinaExpendedora()
maquina.agregar_producto(Producto("Coca Cola", 20, 5))
maquina.agregar_producto(Producto("Sabritas", 15, 3))
maquina.agregar_producto(Producto("Agua", 10, 4))
maquina.agregar_producto(Producto("Chocolate", 25, 2))

while True:
    maquina.mostrar_productos()

    opcion = input("\n¿Qué producto quieres? (número) o 'salir' para terminar: ")

    if opcion == "salir":
        print("¡Hasta luego!")
        break

    dinero = float(input("¿Cuánto dinero ingresas? $"))
    maquina.comprar(int(opcion), dinero)
```
