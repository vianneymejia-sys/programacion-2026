```python
letra_a_puntos = {
    "A": 4.0,
    "B": 3.0,
    "C": 2.0,
    "D": 1.0,
    "F": 0.0
}

puntos_a_letra = {
    4.0: "A",
    3.0: "B",
    2.0: "C",
    1.0: "D",
    0.0: "F"
}
```

```python

while True:
    entrada = input("Ingresa una calificación (o Enter para salir): ")


    if entrada == "":
        break


    try:
        puntos = float(entrada)
        letra = puntos_a_letra[puntos]
        print(f"{puntos} equivale a {letra}")

    except:

        try:
            letra = entrada.upper()
            puntos = letra_a_puntos[letra]
            print(f"{letra} equivale a {puntos}")

        except:

            print(f"'{entrada}' no es una calificación válida")
```
