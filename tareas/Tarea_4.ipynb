```python
# Clase base
class Expresion:
    def evaluar(self):
        raise NotImplementedError("Cada subclase debe implementar evaluar()")


# Representa un número simple
class Numero(Expresion):
    def __init__(self, valor):
        self.valor = valor

    def evaluar(self):
        return self.valor


# Clase intermedia
class OperacionBinaria(Expresion):
    def __init__(self, izq, der):

        if isinstance(izq, (int, float)):
            izq = Numero(izq)
        if isinstance(der, (int, float)):
            der = Numero(der)
        self.izq = izq
        self.der = der

class Suma(OperacionBinaria):
    def evaluar(self):
        return self.izq.evaluar() + self.der.evaluar()


class Resta(OperacionBinaria):
    def evaluar(self):
        return self.izq.evaluar() - self.der.evaluar()


class Producto(OperacionBinaria):
    def evaluar(self):
        return self.izq.evaluar() * self.der.evaluar()
```

```python
print(Suma(3, 4).evaluar())
print(Resta(10, 3).evaluar())
print(Producto(2, 5).evaluar())

print(Suma(Producto(2, 3), 4).evaluar())
```
